```python
shopping_list=[
    {"item": "Milk","price": 50},
    {"item": "Bread","price": 30},
    {"item": "Eggs","price": 60},
    {"item": "Rice","price": 120},
  ]
shopping_list.append({"item": "Butter","price": 80})
removed=shopping_list.pop(0)
print(len(shopping_list))

total = sum(item["price"] for item in shopping_list)

mx=max(shopping_list,key= lambda x:x["price"])
mx_name=mx["item"]
mx_val=mx["price"]
print("Most expensive item:",mx_name,"with price",mx_val)
print("Total cost:",total)

total_items=len(shopping_list)
avg=round(total/total_items,2) if total_items > 0 else 0.0

summary={
  "total_items":total_items,
  "total_cost": total,
  "average_price": avg
}


print("Summary: ",summary)
```
