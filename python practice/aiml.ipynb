Write a simple Python program to evaluate a student's result using conditional statements.

Requirements:
Ask the user to enter:

Student name
Marks in Maths, Science, and English
Validate the marks:

If any mark is less than 0 or greater than 100, print Invalid marks entered and stop the program.
Calculate:

Total marks
Average percentage
Determine Pass / Fail:

If any subject mark is below 40, the student fails
Otherwise, the student passes
If the student passes, assign a grade:

A → Average ≥ 75
B → Average ≥ 60 and < 75
C → Average ≥ 40 and < 60
Display:

Student name
Total marks
Average percentage (2 decimal places)
Pass/Fail status
Grade (only if passed)
Expected Interaction:
Enter student name: Rahul

Enter Maths marks: 78

Enter Science marks: 65

Enter English marks: 82

Student Name: Rahul

Total Marks: 225

Percentage: 75.00

Status: PASS

Grade: A

```python
def main():
    # Prompt for input
    student_name = input("Enter student name: ").strip()

    try:
        maths = int(input("Enter Maths marks: "))
        science = int(input("Enter Science marks: "))
        english = int(input("Enter English marks: "))
    except ValueError:
        print("Invalid marks entered")
        return

    marks = [maths, science, english]

    # Validate marks
    if any(mark < 0 or mark > 100 for mark in marks):
        print("Invalid marks entered")
        return

    # Calculations
    total = sum(marks)
    percentage = total / 3.0

    # Pass/Fail check
    if any(mark < 40 for mark in marks):
        status = "FAIL"
        grade = None
    else:
        status = "PASS"
        # Grade only if passed
        if percentage >= 75:
            grade = "A"
        elif percentage >= 60:
            grade = "B"
        elif percentage >= 40:
            grade = "C"
        else:
            grade = None  # This path won't occur due to prior check

    # Display results
    print(f"Student Name: {student_name}")
    print(f"Total Marks: {total}")
    print(f"Percentage: {percentage:.2f}")
    print(f"Status: {status}")
    if status == "PASS" and grade is not None:
        print(f"Grade: {grade}")

if __name__ == "__main__":
    main()
```
