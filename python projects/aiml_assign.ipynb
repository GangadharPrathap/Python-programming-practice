# Task 1: Create and Write Student Data (Basic)

Approach:

i have Open the file students.txt in write mode to create/overwrite.
Write each "Name,Score" line.
Ensure each record is on its own line.
Close the file (handled by explicit close since using plain open).

```python
filename = "students.txt"

# Data to write
records = [
    "Alice,85",
    "Bob,92",
    "Charlie,78",
    "Diana,95"
]

# Open in write mode to create/overwrite the file
f = open(filename, "w", encoding="utf-8")
for r in records:
    f.write(r + "\n")
f.close()
```

# Task 2: Read and Display Data (Basic)
Approach:

Try to open students.txt and read all lines.
For each line, split on the comma to get name and score.
Print in the required format: Student: [Name], Score: [Score]
If the file doesn't exist, handle the exception gracefully by telling that the file does not exist

```python

filename = "students.txt"

try:
    with open(filename, "r", encoding="utf-8") as f:
        lines = f.readlines()
    for line in lines:
        line = line.strip()
        if not line:
            continue
        if "," in line:
            name, score = line.split(",", 1)
            print(f"Student: {name}, Score: {score}")
        else:
            print(f"Invalid record: {line}")
except FileNotFoundError:
    print(f"File '{filename}' not found.")
```

# Task 3: Append New Student and Create Log (Basic)
Approach:

Append the new student "Eve,88" to the existing students.txt using append mode.
After appending, create a log file activity.log and write the message "Added new student: Eve".

```python
data_filename = "students.txt"
log_filename = "activity.log"
# 1) Append the new student to the existing file
new_record = "Eve,88"
with open(data_filename, "a", encoding="utf-8") as f:
    f.write(new_record + "\n")  # append mode
# 2) Create a log entry
log_message = f"Added new student: Eve"
with open(log_filename, "a", encoding="utf-8") as log_file:
    log_file.write(log_message + "\n")
```
